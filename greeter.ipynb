```python
%%writefile greeter.py
import sys

import argparse

# 1. Set up the parser
parser = argparse.ArgumentParser(description="A script that greets you!")

# 2. Add arguments
parser.add_argument("--name", type=str, default="Stranger", help="Your name")
parser.add_argument("--shout", action="store_true", help="Print in ALL CAPS")
parser.add_argument("--quiet", action="store_true", help="Print in all lowercase")
parser.add_argument("--add", action="store_true", help="Add two numbers")
parser.add_argument("--sub", action="store_true", help="Subtract two numbers")

# 3. Parse the arguments
args = parser.parse_args()

# 4. Use the arguments in the program
message = f"Hello there, {args.name}!"

if args.shout:
    print(message.upper())
else:
    print(message)

if args.quiet:
  print(message.lower())
else:
  print(message)

  sys.argv = [2,3]

if args.add:
  print(int(sys.argv[0]) + int(sys.argv[1]))
else:
  print(message)

```
