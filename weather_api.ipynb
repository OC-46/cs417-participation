```python
import requests

url = "https://api.open-meteo.com/v1/forecast"

params = {
    "latitude":43.05,
    "longitude":-76.15,
    "current_weather": True,
}

city1 = {
    "latitude":60.05,
    "longitude":-40.00,
    "current_weather": True,
}

city2 = {
    "latitude":15.30,
    "longitude":-50.00,
    "current_weather": True,
}

city3 = {
    "latitude":20.00,
    "longitude":-30.00,
    "current_weather": True
}

response = requests.get(url, params=params)
response2 = requests.get(url, params=city1)
response3 = requests.get(url, params=city2)
response4 = requests.get(url, params=city3)
data = response.json()
data2 = response2.json()
data3 = response3.json()
data4 = response4.json()
print(data["current_weather"]["temperature"])
print(data2["current_weather"]["temperature"])
print(data3["current_weather"]["temperature"])
print(data4["current_weather"]["temperature"])


```
